##  local install

```python
import pulp as pl

print("Using PuLP version:", pl.__version__)

# Step 1: Initialize the Model
model = pl.LpProblem("My_First_LP", pl.LpMaximize)

# Step 2: Define Variables directly into the model
x1 = model.add_variable("x1", lowBound=0)  # Product 1
x2 = model.add_variable("x2", lowBound=0)  # Product 2

# Step 3: Define Objective Function
model += 3 * x1 + 2 * x2, "Total_Profit"

# Step 4: Define Constraints
model += 2 * x1 + 1 * x2 <= 100, "Resource_1"
model += x1 + x2 <= 80, "Resource_2"

# Step 5: Solve (using PuLP's default built-in solver)
model.solve()

# Step 6: Print Output
print("x1 =", x1.varValue)
print("x2 =", x2.varValue)
print("Total Profit =", pl.value(model.objective))
```

# Class 2 Prolems

## Two product mix problems

```python
import pulp as pl

# --- 1. Basic Data Arrays (Lists & Dictionaries) ---
products = ["x1", "x2"]
profits = {"x1": 3, "x2": 2}

# Resource availability limits
capacities = {"Resource_1": 100, "Resource_2": 80}

# How much each product uses of Resource 1
r1_usage = {"x1": 2, "x2": 1}

# How much each product uses of Resource 2
r2_usage = {"x1": 1, "x2": 1}


# --- 2. Initialize Model ---
model = pl.LpProblem("Simple_Automation", pl.LpMaximize)


# --- 3. Create Variables in a Dictionary ---
x = {p: model.add_variable(p, lowBound=0) for p in products}


# --- 4. Objective Function ---
model += pl.lpSum([profits[p] * x[p] for p in products]), "Total_Profit"


# --- 5. Constraints ---
# Linking the usage array with the capacity array for Resource 1
model += pl.lpSum([r1_usage[p] * x[p] for p in products]) <= capacities["Resource_1"], "R1_Constraint"

# Linking the usage array with the capacity array for Resource 2
model += pl.lpSum([r2_usage[p] * x[p] for p in products]) <= capacities["Resource_2"], "R2_Constraint"


# --- 6. Solve & Print ---
model.solve()
print({v.name: v.varValue for v in model.variables()})
```

## McDonalds (continuous version)

```python
import pulp as pl

# --- 1. Basic Data Arrays ---
items = ["Hamburger", "Big_Mac", "Chicken_Nuggets", "Garden_Salad", "Apple_Pie"]
prices = {"Hamburger": 0.69, "Big_Mac": 1.99, "Chicken_Nuggets": 1.99, "Garden_Salad": 2.05, "Apple_Pie": 0.79}

calories = {"Hamburger": 260, "Big_Mac": 560, "Chicken_Nuggets": 250, "Garden_Salad": 35, "Apple_Pie": 260}
protein = {"Hamburger": 13, "Big_Mac": 25, "Chicken_Nuggets": 15, "Garden_Salad": 2, "Apple_Pie": 3}
vit_c = {"Hamburger": 2, "Big_Mac": 2, "Chicken_Nuggets": 2, "Garden_Salad": 40, "Apple_Pie": 40}
calcium = {"Hamburger": 15, "Big_Mac": 25, "Chicken_Nuggets": 2, "Garden_Salad": 4, "Apple_Pie": 2}


# --- 2. Initialize Model ---
model = pl.LpProblem("McDonalds_Diet_Continuous", pl.LpMinimize)


# --- 3. Create Variables ---
# cat="Continuous" allows fractions (0.5 burgers). This is the default if left blank.
x = {i: model.add_variable(i, lowBound=0, cat="Continuous") for i in items}


# --- 4. Objective Function ---
model += pl.lpSum([prices[i] * x[i] for i in items]), "Total_Cost"


# --- 5. Constraints ---
model += pl.lpSum([calories[i] * x[i] for i in items]) <= 2000, "Max_Calories"
model += pl.lpSum([protein[i] * x[i] for i in items]) >= 50, "Min_Protein"
model += pl.lpSum([vit_c[i] * x[i] for i in items]) >= 100, "Min_Vit_C"
model += pl.lpSum([calcium[i] * x[i] for i in items]) >= 100, "Min_Calcium"


# --- 6. Solve & Print ---
model.solve()

print({v.name: v.varValue for v in model.variables()})

print("--- Shadow Prices & Slack ---")
for c in model.constraints():
    print(f"{c.name}: Shadow Price = {c.pi}, Slack = {c.slack}")
```

## McDonalds (integer version)

```python
import pulp as pl

items = ["Hamburger", "Big_Mac", "Chicken_Nuggets", "Garden_Salad", "Apple_Pie"]
prices = {"Hamburger": 0.69, "Big_Mac": 1.99, "Chicken_Nuggets": 1.99, "Garden_Salad": 2.05, "Apple_Pie": 0.79}

calories = {"Hamburger": 260, "Big_Mac": 560, "Chicken_Nuggets": 250, "Garden_Salad": 35, "Apple_Pie": 260}
protein = {"Hamburger": 13, "Big_Mac": 25, "Chicken_Nuggets": 15, "Garden_Salad": 2, "Apple_Pie": 3}
vit_c = {"Hamburger": 2, "Big_Mac": 2, "Chicken_Nuggets": 2, "Garden_Salad": 40, "Apple_Pie": 40}
calcium = {"Hamburger": 15, "Big_Mac": 25, "Chicken_Nuggets": 2, "Garden_Salad": 4, "Apple_Pie": 2}

model = pl.LpProblem("McDonalds_Diet_Integer", pl.LpMinimize)

# MAGIC STEP: cat="Integer" forces whole numbers only!
x = {i: model.add_variable(i, lowBound=0, cat="Integer") for i in items}

model += pl.lpSum([prices[i] * x[i] for i in items]), "Total_Cost"

model += pl.lpSum([calories[i] * x[i] for i in items]) <= 2000, "Max_Calories"
model += pl.lpSum([protein[i] * x[i] for i in items]) >= 50, "Min_Protein"
model += pl.lpSum([vit_c[i] * x[i] for i in items]) >= 100, "Min_Vit_C"
model += pl.lpSum([calcium[i] * x[i] for i in items]) >= 100, "Min_Calcium"

model.solve()
print({v.name: v.varValue for v in model.variables()})
```

```python
import pulp as pl

# --- 1. Basic Data Arrays ---
gear = ["Tent", "Stove", "Water_Filter", "Sleeping_Bag"]
values = {"Tent": 10, "Stove": 40, "Water_Filter": 30, "Sleeping_Bag": 50}
weights = {"Tent": 5, "Stove": 4, "Water_Filter": 6, "Sleeping_Bag": 7}

weight_limit = 10

# --- 2. Initialize Model ---
model = pl.LpProblem("Hiking_Knapsack", pl.LpMaximize)


# --- 3. Create Variables ---
# MAGIC STEP: cat="Binary" forces variables to act as Yes/No switches!
x = {g: model.add_variable(g, cat="Binary") for g in gear}


# --- 4. Objective Function ---
model += pl.lpSum([values[g] * x[g] for g in gear]), "Total_Value"


# --- 5. Constraints ---
model += pl.lpSum([weights[g] * x[g] for g in gear]) <= weight_limit, "Max_Weight"


# --- 6. Solve & Print ---
model.solve()
print({v.name: v.varValue for v in model.variables()})
```

## Discussion Problem

```python
import pulp as pl

# --- 1. Basic Data Arrays ---
drinks = ["Espresso", "Latte"]
profits = {"Espresso": 3, "Latte": 4}

beans_used = {"Espresso": 2, "Latte": 1}
milk_used = {"Espresso": 1, "Latte": 2}

# --- 2. Initialize Model ---
# We want to MAXIMIZE profit
model = pl.LpProblem("Coffee_Shop_Optimization", pl.LpMaximize)

# --- 3. Create Variables ---
# We must sell whole drinks, so we use cat="Integer"
x = {d: model.add_variable(d, lowBound=0, cat="Integer") for d in drinks}

# --- 4. Objective Function ---
model += pl.lpSum([profits[d] * x[d] for d in drinks]), "Total_Profit"

# --- 5. Constraints ---
model += pl.lpSum([beans_used[d] * x[d] for d in drinks]) <= 100, "Max_Beans"
model += pl.lpSum([milk_used[d] * x[d] for d in drinks]) <= 80, "Max_Milk"

# --- 6. Solve & Print ---
model.solve()

print("--- Optimal Daily Menu ---")
for d in drinks:
    print(f"{d}s to make = {x[d].varValue}")
    
print(f"Total Projected Profit = ${pl.value(model.objective)}")
```
